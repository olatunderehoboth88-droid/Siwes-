# Fifth Assignment

```python
import pandas as pd
```

```python
def stud_grade():
    names=[]
    Mathematics=[]
    English=[]
    Basic_Science=[]
    Crs=[]
    Computer_studies=[]
    Economics=[]
    History=[]
    Social_studies=[]
    civic_education=[]
    student_no=int(input('Enter number of students:'))
    for i in range(student_no):
        name=input('Enter Student full name:')
        mth=int(input('What is your Mathematics Score:'))
        eng=int(input('What is your English Score:'))
        bsc=int(input('What is your Basic Science Score:'))
        crs=int(input('What is your CRS Score:'))
        csc=int(input('What is your Computer Studies Score:'))
        ec=int(input('What is your economics Score:'))
        his=int(input('What is your History Score:'))
        scs=int(input('What is your Social Studies Score:'))
        ce=int(input('What is your Civic Education Score:'))
        names.append(name)
        Mathematics.append(mth)
        English.append(eng)
        Basic_Science.append(bsc)
        Crs.append(crs)
        Computer_studies.append(csc)
        Economics.append(ec)
        History.append(his)
        Social_studies.append(scs)
        civic_education.append(ce)
    dictionary={
        'Name': names,
        'Mathematics': Mathematics,
        'English':English,
        'Basic Science':Basic_Science,
        'CRS':Crs,
        'Computer studies':Computer_studies,
        'Economics':Economics,
        'History':History,
        'Social_studies':Social_studies,
        'Civic_education': civic_education
    }
    df=pd.DataFrame(dictionary)
    df['Total_avg']=(df['Mathematics']+df['English']+df['Basic Science']+df['CRS']+df['Computer studies']+df['Economics']+df['History']+df['Social_studies']+df['Civic_education'])/9
    grade=[]
    for i in df['Total_avg']:
        if i>=70:
            grade.append('A')
        elif i<70 and i>=60:
            grade.append('B')
        elif i<60 and i>=50:
            grade.append('C')
        elif i<50 and i>=45:
            grade.append('D')
        elif i<45 and i>39:
            grade.append('E')
        elif i<=39:
            grade.append('F')
        else:
            print('Invalid Score')
    df['Grade']= grade
    print('Class Result:')
    print('='*60)
    print(df) 
```

# Sixth Assignment

```python
import pandas as pd
import os
import datetime as datetime
```

```python
GRADE_FILE='grade.csv'
def year_grade():
    if os.path.exists(GRADE_FILE):
        df= pd.read_csv(GRADE_FILE)
        for col in ['Name','Subject Name','First Term','Second Term','Third Term']:
           if col not in df.columns:
            df [col] = []
    else:
            df = pd.DataFrame(columns= ['Name','Subject Name','First Term','Second Term','Third Term'])
    if not df.empty:
            df['Total_avg']=(df['First Term']+df['Second Term']+df['Third Term'])/3
            grade=[]
            for i in df['Total_avg']:
                if i>=70:
                    grade.append('A')
                elif i<70 and i>=60:
                    grade.append('B')
                elif i<60 and i>=50:
                    grade.append('C')
                elif i<50 and i>=45:
                    grade.append('D')
                elif i<45 and i>39:
                    grade.append('E')
                elif i<=39:
                    grade.append('F')
                else:
                    print('Invalid Score')
            df['Grade']= grade
    return df
def save_inventory(df):
    df.to_csv(GRADE_FILE, index=False)
    print("Inventory saved successfully.")
def display_inventory(df):
    if df.empty:
        print("\nEmpty Records.")
        return
    print("\n" + "=" * 70)
    print(df.to_string(index=False))
def add_record(df):
    name = input("Enter student name: ").strip()
    if not name:
        print("Student name cannot be empty.")
        return df
    existing = df[df['Name'].str.lower() == name.lower()]
    if not existing.empty:
        print(f'Student Record already exist, remove {name} and start again')
    else:
        try:
            subj_name = input(f"Enter subject name, {name}: ")
            ft = int(input(f"Enter First Term score, {name}: "))
            st = int(input(f"Enter Second Term score, {name}: "))
            lt = int(input(f"Enter Last Term score, {name}: "))
            new_row = pd.DataFrame({
                'Name': [name],
                'Subject Name':[subj_name],
                'First Term': [ft],
                'Second Term': [st],
                'Third Term':[lt]
            })
            df = pd.concat([df, new_row], ignore_index=True)
            print(f"Added {name}'s Record to inventory.")
        except ValueError:
            print("Invalid Value.")
    df['Total_avg']=(df['First Term']+df['Second Term']+df['Third Term'])/3
    grade=[]
    for i in df['Total_avg']:
        if i>=70:
            grade.append('A')
        elif i<70 and i>=60:
            grade.append('B')
        elif i<60 and i>=50:
            grade.append('C')
        elif i<50 and i>=45:
            grade.append('D')
        elif i<45 and i>39:
            grade.append('E')
        elif i<=39:
            grade.append('F')
        else:
            print('Invalid Score')
    df['Grade']= grade
    return df
def remove_record(df):
    if df.empty:
        print("Student Record is empty.")
        return df
    display_inventory(df)
    name = input("Enter Student name to remove: ").strip()
    match = df[df['Name'].str.lower() == name.lower()]
    if match.empty:
        print("Student not found.")
        return df
    df = df.drop(match.index).reset_index(drop=True)
    print(f"Removed {name}'s record Succesfully'.")
    return df
def main():
    df = year_grade()
    while True:
        print("\n" + "=" * 40)
        print("       INVENTORY MANAGER")
        print("=" * 40)
        print("1. View Student Records")
        print("2. Add Student Record")
        print("3. Remove Student Record")
        print("4. Save & Exit")
        print("0. Exit without saving")
        choice = input("\nChoose an option: ").strip()
        if choice == '1':
            display_inventory(df)
        elif choice == '2':
            df= add_record(df)
        elif choice== '3':
            df= remove_record(df)
        elif choice== '4':
            save_inventory(df)
            print("Goodbye!")
            break
        elif choice == '0':
            print("Exited without saving.")
            break
        else:
            print("Invalid option.")
if __name__ == "__main__":
    main()
```
