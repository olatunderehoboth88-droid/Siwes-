```python
import pandas as pd
```

```python
def gen_reciepts():
    prod_name=[]
    qty=[]
    prices=[]
    name=input('Enter your name:')
    num_prod=int(input('Enter numbers of products'))
    for i in range(num_prod):
        if i<=num_prod:
           prod=input(f'enter product {i+1} name')
           q= int(input(f'enter the quantity of {prod}'))
           price=float(input(f'enter price of {prod}'))
           prod_name.append(prod)
           qty.append(q)
           prices.append(price)
        diction={
            'prod_name': prod_name,
            'qty': qty,
            'prices': prices
        }
    df= pd.DataFrame(diction)
    df['Amount']= df['qty']*df['prices']
    df['Discounted']= df['Amount']*0.5
    print('='*60)
    print(df)
```

### class work

```python
def year_grade():
    subject_name=[]
    first_term=[]
    second_term=[]
    third_term=[]
    name= input('Enter your name:')
    subj_no=int(input('Enter subjects numbers'))
    for i in range(subj_no):
        if i<=(subj_no):
            subj_name=input('Enter subject name: ')
            ft=int(input('Enter first term score: '))
            st= int(input('Enter second term score: '))
            tt= int(input('Enter third term score: '))
            subject_name.append(subj_name)
            first_term.append(ft)
            second_term.append(st)
            third_term.append(tt)
    dictionary={
        'Subject_name': subject_name,
        'First_term': first_term,
        'Second_term': second_term,
        'Third_term': third_term
    }        
    df=pd.DataFrame(dictionary)
    df['Total_avg']=(df['First_term']+df['Second_term']+df['Third_term'])/3
    grade=[]
    for i in df['Total_avg']:
        if i>=70:
            grade.append('A')
        elif i<70 and i>=60:
            grade.append('B')
        elif i<60 and i>=50:
            grade.append('C')
        elif i<50 and i>=45:
            grade.append('D')
        elif i<45 and i>39:
            grade.append('E')
        elif i<=39:
            grade.append('F')
        else:
            print('Invalid Score')
    df['Grade']= grade
    print(name, 'Year Reult:')
    print('='*60)
    print(df)
```

```python
year_grade()
```
